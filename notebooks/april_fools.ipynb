```python
from dotenv import load_dotenv

import os
import requests

load_dotenv()
```

```python
discord_url = os.environ.get('APRIL_FOOLS_DISCORD_WEBHOOK_URL')
```

```python
def send(msg):
    payload = {"content": msg}
    try:
        requests.post(discord_url, json=payload)
    except Exception as e:
        print(f"ERROR: Failed to send Discord message: {e}")
```
